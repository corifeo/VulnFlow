# Complete Library
Build a complete NVD vulnerability database (2002-present).

```python
from vulnflow.helpers import CacheBuilder

ds = CacheBuilder().build()
ds.to_parquet("../data/complete_library.parquet")
print(f"Saved {len(ds):,} CVEs to data/complete_library.parquet")
```

## Overview

```python
df = ds.to_pandas()
print(f"{len(df):,} CVEs  |  {len(df.columns)} columns")
print(f"Year range: {df['cve_id'].str[4:8].min()}-{df['cve_id'].str[4:8].max()}")
```

## Severity Distribution

```python
df["cvss_base_severity"].value_counts()
```

## CVEs per Year

```python
df["year"] = df["cve_id"].str[4:8].astype(int)
yearly = df.groupby("year").agg(
    count=("cve_id", "size"),
    mean_score=("cvss_base_score", "mean"),
).round(1)
yearly.tail(10)
```

## NVD Status Breakdown

```python
# to_pandas() excludes 'rejected' by default
# Use rejected=True to include them
df_all = ds.to_pandas(rejected=True)
df_all["nvd_status"].value_counts()
```

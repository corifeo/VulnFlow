# Complete Library
Build a complete NVD vulnerability database (2002-present).

Adjust the year range below to control download size.

```python
# Configuration - adjust as needed
YEARS = [2023, 2024, 2025]  # Change to None for ALL years (2002-present)
WORKERS = 4
```

```python
from vulnflow.helpers import CacheBuilder

if YEARS:
    builder = CacheBuilder(years=YEARS, include_modified=True)
else:
    builder = CacheBuilder(include_modified=True)

ds = builder.build(workers=WORKERS)
print(f"{len(ds)} CVEs loaded")
```

```python
ds.to_parquet("../data/complete_library.parquet")
print("Saved to data/complete_library.parquet")
```

## Explore the Library

```python
df = ds.to_pandas()
print(f"Records: {len(df)}")
print(f"Columns: {len(df.columns)}")
print(f"\nYear range: {df['cve_id'].str[4:8].min()}-{df['cve_id'].str[4:8].max()}")
df["cvss_base_severity"].value_counts()
```

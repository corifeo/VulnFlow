# Quick Exploration
Explore VulnFlow with the NVD Modified feed (last 8 days, ~2-5K CVEs).

```python
from vulnflow.helpers import CacheBuilder
ds = CacheBuilder(modified_only=True).build(workers=1)
```

```python
df = ds.to_pandas()
print(f"{len(df)} CVEs loaded")
df.head()
```

## Severity Distribution

```python
df["cvss_base_severity"].value_counts()
```

## Critical Vulnerabilities

```python
critical = df[df["cvss_base_score"] >= 9.0]
print(f"{len(critical)} critical CVEs")
critical[["cve_id", "cvss_base_score", "cvss_base_severity"]].head(10)
```

## Save for Later
Cache to parquet for fast reloading.

```python
ds.to_parquet("../data/quick_cache.parquet")
print("Saved to data/quick_cache.parquet")
```

# Quick Exploration
Explore VulnFlow with the NVD Modified feed (last 8 days, ~2-5K CVEs).

```python
from vulnflow.helpers import CacheBuilder

ds = CacheBuilder(modified_only=True).build()
df = ds.to_pandas()
print(f"{len(df):,} CVEs  |  {len(df.columns)} columns")
df.head(3)
```

## Severity Distribution

```python
df["cvss_base_severity"].value_counts()
```

## Critical Vulnerabilities

```python
critical = df[df["cvss_base_score"] >= 9.0]
print(f"{len(critical):,} critical CVEs")
critical[["cve_id", "cvss_base_score", "cvss_attack_vector", "description"]].head(10)
```

## Network-Accessible, High-Impact

```python
network_critical = df[
    (df["cvss_attack_vector"] == "NETWORK")
    & (df["cvss_base_score"] >= 9.0)
    & (df["cvss_privileges_required"] == "NONE")
]
print(f"{len(network_critical):,} unauthenticated network-critical CVEs")
network_critical[["cve_id", "cvss_base_score", "description"]].head(5)
```

## Save for Later

```python
ds.to_parquet("../data/quick_cache.parquet")
print("Saved to data/quick_cache.parquet")
```
